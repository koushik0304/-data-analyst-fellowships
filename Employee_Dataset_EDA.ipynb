Dataset: `Employee_Dataset.csv` — 50 employee records with blood group, age, healthy eating score, active lifestyle score, and salary.

## 1. Setup & Load Data

```python
df = pd.read_csv('Employee_Dataset.csv')
df.head()
```

## 4. Categorical Feature: `groups`

```python
df['groups'].value_counts()
```

```python
plt.figure(figsize=(8,5))
sns.countplot(x='groups', data=df, order=df['groups'].value_counts().index)
plt.title('Count by Blood Group')
plt.show()
```

## 5. Outlier Detection (IQR method)

```python
numeric_cols = ['age', 'healthy_eating', 'active_lifestyle', 'salary']
```

```python
Q1 = df[numeric_cols].quantile(0.25)
Q3 = df[numeric_cols].quantile(0.75)
IQR = Q3 - Q1

outlier_mask = ((df[numeric_cols] < (Q1 - 1.5*IQR)) | (df[numeric_cols] > (Q3 + 1.5*IQR))).any(axis=1)
df_no_outliers = df[~outlier_mask]

print("Original shape:", df.shape)
print("After removing outliers:", df_no_outliers.shape)
print("Outlier rows removed:", outlier_mask.sum())
```

```python
plt.figure(figsize=(10,6))
sns.boxplot(data=df[numeric_cols])
plt.title('Boxplots of Numeric Columns')
plt.show()
```

```python
plt.figure(figsize=(8,5))
sns.boxplot(x=df['salary'])
plt.title('Boxplot of Salary')
plt.show()
```

`salary` has the widest spread and shows a couple of high-end outliers via the IQR rule; `age`, `healthy_eating`, and `active_lifestyle` are comparatively tight.

## 6. Univariate Distributions

```python
plt.figure(figsize=(15,10))
for i, col in enumerate(numeric_cols, 1):
    plt.subplot(2, 2, i)
    sns.histplot(df[col], kde=True)
    plt.title(f'Distribution of {col}')
plt.tight_layout()
plt.show()
```

## 7. Bivariate Analysis

```python
plt.figure(figsize=(8,5))
sns.boxplot(x='groups', y='salary', data=df)
plt.title('Salary by Group')
plt.show()
```

```python
plt.figure(figsize=(8,5))
sns.scatterplot(x='age', y='salary', hue='groups', data=df)
plt.title('Age vs Salary')
plt.show()
```

```python
plt.figure(figsize=(8,5))
sns.scatterplot(x='healthy_eating', y='active_lifestyle', hue='groups', data=df)
plt.title('Healthy Eating vs Active Lifestyle')
plt.show()
```

```python
plt.figure(figsize=(8,5))
sns.scatterplot(x='active_lifestyle', y='salary', data=df)
plt.title('Active Lifestyle vs Salary')
plt.show()
```

## 8. Correlation Analysis

```python
corr = df[numeric_cols].corr()
corr
```

```python
plt.figure(figsize=(8,6))
sns.heatmap(corr, annot=True, cmap='coolwarm')
plt.title('Correlation Heatmap')
plt.show()
```

## 9. Pairplot — All Numeric Features by Group

```python
g = sns.pairplot(df[numeric_cols + ['groups']], hue='groups')
g.fig.suptitle('Pairplot of Numeric Features by Group', y=1.02)
plt.show()
```

## 10. Summary

- **50 records, 6 columns**, no missing values, no duplicates.
- `groups` (blood type) has 4 categories: A, O, AB, B.
- `salary` ranges from **556 to 4969** with a mean around **2028**; the IQR method flags a small number of high-salary outliers.
- Correlation values between `age`, `healthy_eating`, `active_lifestyle`, and `salary` are visible in the heatmap above — use these to judge whether lifestyle factors track with salary or age in this sample.
